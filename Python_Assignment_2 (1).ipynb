## List
### 1. List Creation

```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Shihana", "Janani", "Arun", "Meera", "Rahul"]

print("age_list:", age_list)
print("name_list:", name_list)
```

### 2. List Operations / Modifications

```python
# a Append "Yazhini" to name_list
name_list.append("Yazhini")
print("a. After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("b. After insert:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("c. After remove:", name_list)

# d. Pop the last element from age_list
popped = age_list.pop()
print("d. Popped element:", popped, "| age_list:", age_list)

# e. Extend age_list with [29, 30, 26]
age_list.extend([29, 30, 26])
print("e. After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("f. Sorted (descending):", age_list)

# g. Max, Min and Sum
print("g. Max age:", max(age_list))
print("   Min age:", min(age_list))
print("   Sum of ages:", sum(age_list))
```

### 3. Accessing List Elements

```python
print("a. First element:", name_list[0])
print("b. Last element:", name_list[-1])
print("c. Index 2 to 4:", name_list[2:5])
print("d. Reverse order:", name_list[::-1])
```

## Dictionary

```python
# a. Create the dictionary
student_marks = {
    "Shihana": 92,
    "Arun": 78,
    "Meera": 85,
    "Rahul": 69,
    "Divya": 74
}
print("a. student_marks:", student_marks)

# b. Access the mark of a specific student
print("b. Marks of Meera:", student_marks["Meera"])

# c. Add a new student "Janani" with 80
student_marks["Janani"] = 80
print("c. After adding Janani:", student_marks)

# d. Update the mark of an older student to 82
student_marks["Rahul"] = 82
print("d. After updating Rahul:", student_marks)

# e. keys(), values(), items()
print("e. Keys:", student_marks.keys())
print("   Values:", student_marks.values())
print("   Items:", student_marks.items())
```

## Sets

### a. Creating `my_set` from a list with duplicate values

```python
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
```

Explanation: A set stores only unique elements, so the repeated `'a'` and `'i'` values are removed automatically. The output contains just the five vowels `a, e, i, o, u`. A set is also unordered, so the elements may be printed in a different order each time you run it.

### b. Trying to change `my_set[4] = 's'`

```python
try:
    my_set[4] = 's'
except TypeError as e:
    print("TypeError:", e)
```

**Explanation:** This throws a TypeError because sets are **unordered** and do not support indexing or item assignment. Since elements have no fixed position, my_set[4] has no meaning.

### c. & d. Two sets, union and intersection

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("set1:", set1)
print("set2:", set2)
print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
```
